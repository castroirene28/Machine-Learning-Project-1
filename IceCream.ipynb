````python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt


# Ice Cream Flavor Samples

flavors = [
    'Vanilla', 'Chocolate', 'Strawberry', 'Mint',
    'Vanilla', 'Chocolate', 'Vanilla', 'Cookies & Cream',
    'Chocolate', 'Strawberry', 'Vanilla', 'Mint',
    'Chocolate', 'Vanilla', 'Rocky Road', 'Chocolate',
    'Cookies & Cream', 'Vanilla', 'Strawberry', 'Mint',
    'Vanilla', 'Chocolate', 'Vanilla', 'Chocolate',
    'Mint', 'Vanilla', 'Chocolate', 'Strawberry',
    'Cookies & Cream', 'Vanilla'
]

# Create DataFrame

df = pd.DataFrame({'Flavor': flavors})

# Count flavor popularity
flavor_counts = df['Flavor'].value_counts()

# Display popularity counts
print("Ice Cream Flavor Popularity:\n")
print(flavor_counts)

# ----------------------------
# Color palette
# ----------------------------
colors = {
    'Vanilla': '#F3E5AB',          # Vanilla cream
    'Chocolate': '#7B3F00',        # Chocolate brown
    'Strawberry': '#FF6384',       # Pink
    'Mint': '#98FF98',             # Mint green
    'Cookies & Cream': '#D3D3D3',  # Light gray
    'Rocky Road': '#8B4513'        # Dark brown
}


# Bar Chart

bar_colors = [colors[flavor] for flavor in flavor_counts.index]

plt.figure(figsize=(8, 5))
plt.bar(
    flavor_counts.index,
    flavor_counts.values,
    color=bar_colors,
    edgecolor='black'
)

plt.title('Popularity of Ice Cream Flavors')
plt.xlabel('Ice Cream Flavor')
plt.ylabel('Number of Votes')
plt.xticks(rotation=30)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()


# Histogram

df['Flavor_Code'] = df['Flavor'].astype('category').cat.codes
categories = df['Flavor'].astype('category').cat.categories

plt.figure(figsize=(8, 5))

counts, bins, patches = plt.hist(
    df['Flavor_Code'],
    bins=len(categories),
    edgecolor='black'
)

hist_colors = [colors.get(category, '#87CEEB') for category in categories]

for patch, color in zip(patches, hist_colors):
    patch.set_facecolor(color)

plt.xticks(
    range(len(categories)),
    categories,
    rotation=30
)

plt.title('Histogram of Ice Cream Flavors')
plt.xlabel('Ice Cream Flavor')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()


# Pie Chart

pie_colors = [colors[flavor] for flavor in flavor_counts.index]

plt.figure(figsize=(7, 7))

plt.pie(
    flavor_counts.values,
    labels=flavor_counts.index,
    colors=pie_colors,
    autopct='%1.1f%%',
    startangle=90,
    counterclock=False,
    wedgeprops={'edgecolor': 'black'}
)

plt.title('Ice Cream Flavor Popularity (%)')
plt.axis('equal')  # Makes the pie chart a circle
plt.tight_layout()
plt.show()
````

````python
percentage_share = (flavor_counts / flavor_counts.sum()) * 100
print("Ice Cream Flavor Percentage Share:\n")
print(percentage_share.round(2))
````
